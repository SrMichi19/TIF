## 1. INTRODUCCIÓN

El análisis de señales fisiológicas es una herramienta clave para el estudio y comprensión de la actividad eléctrica de distintos sistemas biológicos, como el cerebro, el corazón y los músculos. Estas señales, captadas mediante técnicas como la electroencefalografía (EEG), la electrocardiografía (ECG) y la electromiografía (EMG), proporcionan información esencial tanto en el ámbito clínico como en el de investigación. A través del uso de electrodos superficiales, es posible registrar la actividad bioeléctrica de estos sistemas y analizarla en función de parámetros específicos, permitiendo detectar patologías, evaluar el funcionamiento fisiológico y desarrollar nuevas tecnologías de monitoreo y diagnóstico.

El procesamiento de estas señales resulta fundamental, ya que frecuentemente se encuentran contaminadas por ruidos o artefactos que dificultan su interpretación. Por esta razón, es necesario aplicar técnicas computacionales que permitan filtrar, visualizar y extraer características relevantes de los registros obtenidos. Este tipo de análisis se encuentra cada vez más automatizado gracias al desarrollo de librerías y herramientas basadas en lenguajes de programación como Python.

El presente informe tiene como objetivo el diseño e implementación de una librería en Python para el procesamiento y análisis básico de señales fisiológicas EEG, ECG y EMG. Se utilizarán recursos como NumPy, Pandas y Matplotlib, en combinación con los principios de la Programación Orientada a Objetos (POO), para desarrollar un conjunto modular y escalable de clases y funciones. Asimismo, se realizarán ejemplos de uso y pruebas en Jupyter Notebook, integrando técnicas de filtrado, detección de eventos y visualización gráfica, simulando un entorno de aplicación clínica o experimental.

## 2. MARCO TEÓRICO

El estudio y tratamiento de señales fisiológicas representan un área esencial dentro de la ingeniería biomédica, la neurociencia y la medicina clínica. Estas señales reflejan la actividad eléctrica de distintas bioseñales del cuerpo humano y constituyen una herramienta fundamental para la detección, el seguimiento y la investigación de condiciones fisiológicas y patológicas.

**Entre las señales más relevantes desde el punto de vista clínico y experimental se destacan:**
* **Electroencefalograma (EEG):** Es un método de registro no invasivo que permite captar la actividad bioeléctrica cerebral mediante sensores colocados sobre el cuero cabelludo. Se utiliza para evaluar el funcionamiento neuronal en casos de epilepsia, alteraciones del sueño y también en contextos de neurociencia cognitiva. El EEG se compone de diferentes bandas de frecuencia —delta, theta, alfa y beta—, cada una asociada a distintos niveles de actividad cerebral.
* **Electrocardiograma (ECG):** Registra las señales eléctricas del corazón que resultan de la despolarización y repolarización del músculo cardíaco. Es un recurso diagnóstico de primer orden para detectar trastornos del ritmo, isquemias o eventos cardiovasculares agudos. Las características del trazado ECG permiten analizar en detalle eventos como la conducción auriculoventricular y el estado funcional del miocardio .
* **Electromiograma (EMG):** Permite medir la actividad eléctrica generada por los músculos durante la contracción. Esta señal proporciona información clave sobre la eficiencia neuromuscular, el estado de fatiga y la fuerza aplicada. Su aplicación es habitual en estudios de biomecánica, control motor, fisioterapia y desarrollo de prótesis inteligentes.

Las señales fisiológicas suelen estar afectadas por diferentes tipos de interferencias, tanto externas como internas (ruido eléctrico, artefactos de movimiento, entre otros), lo que dificulta su análisis directo. Por esta razón, es necesario aplicar técnicas de procesamiento digital que permitan limpiar, resaltar y extraer información significativa de los datos obtenidos.

**Propósitos del Procesamiento de Señales Biomédicas**
* **Reducción de interferencias:** Se aplican filtros digitales (como pasa bajos, pasa altos o de rechazo de banda) y métodos más sofisticados como la separación de componentes independientes, con el fin de eliminar distorsiones no deseadas.
* **Identificación de eventos fisiológicos:** Por ejemplo, la localización de picos R en ECG permite calcular parámetros cardíacos; en EEG, la observación de ondas alfa permite reconocer estados de reposo o concentración; en EMG, los patrones de activación indican niveles de esfuerzo muscular.
* **Análisis cuantitativo:** Se extraen parámetros tanto del dominio temporal (promedio, desviación) como del espectro de frecuencias (mediante la transformada de Fourier), permitiendo evaluar ritmos cerebrales, variabilidad cardíaca o contenido energético de la señal muscular.
* **Representación gráfica:** La visualización de los datos mediante gráficos, mapas topográficos o espectrogramas facilita el diagnóstico, la comparación entre estados fisiológicos y el seguimiento de la evolución de un paciente.

**Ejemplos de Aplicación**
* **EEG:** En epilepsia, el EEG permite detectar descargas anómalas o ritmos característicos que orientan el diagnóstico.
* **ECG:** En eventos cardíacos, variaciones del segmento ST en el ECG pueden ser signos precoces de infarto.
* **EMG:** En rehabilitación muscular, una disminución de la frecuencia alta en el EMG señala la aparición de fatiga, lo que puede ser útil para ajustar el entrenamiento.
