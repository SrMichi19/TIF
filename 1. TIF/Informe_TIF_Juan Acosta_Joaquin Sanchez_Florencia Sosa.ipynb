## 2. MARCO TEÓRICO

El procesamiento de señales fisiológicas es un campo central dentro de la ingeniería biomédica y las neurociencias, ya que permite estudiar y cuantificar fenómenos eléctricos generados por el cuerpo humano. Las señales fisiológicas representan la actividad bioeléctrica de distintos sistemas del organismo y constituyen una herramienta fundamental para el diagnóstico clínico, la monitorización de pacientes y la investigación científica. Dentro de las más estudiadas y utilizadas se encuentran la electroencefalografía (EEG), la electrocardiografía (ECG) y la electromiografía (EMG).

La señal EEG refleja la actividad eléctrica cerebral captada mediante electrodos dispuestos sobre el cuero cabelludo. Esta señal se caracteriza por tener baja amplitud, estar altamente sujeta a ruido y variar según el estado cognitivo o fisiológico del individuo. El análisis del EEG permite detectar patrones como ondas alfa o beta, identificar eventos epilépticos, y estudiar la dinámica cerebral en diferentes contextos. Por su parte, la señal ECG registra la actividad eléctrica del corazón y es ampliamente utilizada en la clínica para evaluar el ritmo cardíaco, identificar arritmias, isquemias y eventos agudos como el infarto de miocardio. Finalmente, la señal EMG recoge la actividad eléctrica generada por los músculos durante su contracción, lo que permite estudiar el funcionamiento neuromuscular, evaluar patologías o incluso controlar dispositivos protésicos.

Estas señales presentan características particulares que hacen necesario un procesamiento cuidadoso antes de ser interpretadas. Por ejemplo, suelen estar contaminadas con diferentes tipos de ruidos (como la interferencia de la red eléctrica a 50/60 Hz) y artefactos fisiológicos o de movimiento (como parpadeos en EEG, desplazamientos de electrodos en ECG, o contracciones involuntarias en EMG). Además, contienen componentes de interés que se encuentran embebidos en una dinámica no lineal y compleja, por lo que el procesamiento digital resulta indispensable para mejorar su calidad y extraer información significativa.

Entre las técnicas más utilizadas se encuentra el filtrado digital, que permite eliminar componentes indeseadas del espectro de la señal. Por ejemplo, mediante filtros notch se elimina la interferencia de la red eléctrica, y con filtros pasabanda se selecciona el rango fisiológico relevante (como 1–40 Hz para EEG). Otra herramienta importante es el análisis espectral, que permite transformar la señal desde el dominio temporal al dominio de la frecuencia mediante la Transformada de Fourier (FFT) o el método de Welch. Esto permite identificar ritmos cerebrales en EEG, analizar la distribución de energía muscular en EMG, o estudiar bandas de potencia cardíaca en ECG.

La detección de eventos es otro aspecto fundamental del análisis de señales fisiológicas. En EEG, por ejemplo, se suele marcar el inicio de estímulos o tareas cognitivas, lo cual facilita el análisis de potenciales evocados. En ECG, la detección de los picos R del complejo QRS es esencial para calcular la frecuencia cardíaca y su variabilidad. En EMG, detectar activaciones musculares permite estimar la intensidad del esfuerzo, la fatiga o la coordinación motora. Para esto, se requiere una estructura de anotaciones flexible y precisa.

Finalmente, la visualización juega un rol clave, tanto en el análisis exploratorio como en la interpretación clínica o científica de las señales. Representar las señales en el tiempo, visualizar anotaciones superpuestas, observar los espectros de potencia o detectar gráficamente la ocurrencia de eventos fisiológicos facilita la toma de decisiones y la comunicación de resultados.

El diseño de una librería que integre estas funcionalidades requiere una estructura modular, robusta y reutilizable. La programación orientada a objetos (POO) en Python permite construir clases específicas para cada tipo de señal (EEG, ECG, EMG), encapsular los métodos de procesamiento y visualización, y garantizar la escalabilidad del software. A su vez, el uso de bibliotecas científicas como NumPy, Pandas, Matplotlib y SciPy asegura un tratamiento eficiente de los datos y una representación visual clara y personalizable.

Internamente, el método llama a uno de tres métodos privados según el tipo de referencia seleccionada. Si se elige `"canal"`, se utiliza `_aplicar_referencia_a_canal`, que resta la señal del canal indicado al resto de los canales. En cambio, si se selecciona `"promedio"`, se invoca `_aplicar_referencia_promedio`, que resta a cada canal el promedio entre todos ellos. Finalmente, si se opta por `"laplaciano"`, se ejecuta `_aplicar_referencia_laplaciana`, que destaca la actividad local de cada canal comparándolo con sus vecinos.

El resultado de esta operación se guarda en el atributo `data_ref`, que almacena la versión referenciada de la señal. En caso de que se ingrese un tipo de referencia inválido, el método lanza un `ValueError`. De este modo, `set_reference` permite adaptar la señal EEG a distintas necesidades analíticas de forma flexible y controlada.

```python
ref = eeg.set_reference(reference="canal", channel="2")
ref.shape
ref[0]
```

Se cambia la referencia de la señal EEG a una referencia *laplaciana*. Al ejecutar `eeg.set_reference("laplaciano")`, se aplica un filtro espacial que resalta la actividad eléctrica local de cada canal, comparándola con la de sus canales vecinos (definidos previamente en el objeto `Info`). Este tipo de referencia es útil para mejorar la localización de fuentes cerebrales activas.

El método interno que utiliza este método es `_aplicar_referencia_laplaciana`, que recorre cada canal, identifica sus vecinos y calcula la diferencia entre la señal del canal y el promedio de sus vecinos. El resultado se almacena en `data_ref` y, al acceder a `ref_laplaciana[0]`, se visualiza la señal laplaciana del primer canal.

```python
ref_laplaciana = eeg.set_reference("laplaciano")
ref_laplaciana[0]
```

Se cambia la referencia de la señal EEG a una referencia **promedio** mediante el método `set_reference(reference="promedio")`. Esto significa que a cada canal se le resta, muestra a muestra, el promedio de todos los canales en ese instante, lo que ayuda a reducir el ruido común compartido entre ellos y mejora la interpretación de la actividad relativa.

Internamente, se utiliza el método `_aplicar_referencia_promedio`, que calcula el promedio por columna (es decir, en el tiempo) y genera una nueva matriz referenciada, almacenada en `data_ref`. Al consultar `ref_promedio.shape`, se obtiene la forma de la matriz referenciada, y `ref_promedio[0]` devuelve la señal del primer canal con esta referencia aplicada.

```python
ref_promedio = eeg.set_reference(reference="promedio")
ref_promedio.shape
ref_promedio[0]
```
